````python

evaluation_results = []

for item in evaluation_data:

    retrieved_docs = retrieve_documents(
        query=item["query"],
        top_k=3,
        intent=item["intent"]
    )

    retrieved_ids = [
        doc["chunk_id"]
        for doc in retrieved_docs
    ]

    expected_id = item["expected_chunk_id"]

    # Precision@1
    precision_at_1 = (
        1.0 if retrieved_ids and retrieved_ids[0] == expected_id
        else 0.0
    )

    # Recall@3
    recall_at_3 = (
        1.0 if expected_id in retrieved_ids
        else 0.0
    )

    # Mean Reciprocal Rank
    if expected_id in retrieved_ids:
        rank = retrieved_ids.index(expected_id) + 1
        reciprocal_rank = 1 / rank
    else:
        rank = None
        reciprocal_rank = 0.0

    evaluation_results.append({
        "query": item["query"],
        "expected_chunk_id": expected_id,
        "retrieved_chunk_ids": retrieved_ids,
        "correct_rank": rank,
        "precision_at_1": precision_at_1,
        "recall_at_3": recall_at_3,
        "reciprocal_rank": reciprocal_rank
    })


results_df = pd.DataFrame(evaluation_results)

print("Retrieval evaluation completed.")
print(f"Evaluated queries: {len(results_df)}")

display(results_df)
````

````python
import pandas as pd

from sklearn.metrics.pairwise import cosine_similarity

from src.retriever import (
    load_retriever,
    INTENT_SEARCH_TERMS
)


# Load the knowledge base retriever
retriever = load_retriever()

vectorizer = retriever["vectorizer"]
document_matrix = retriever["document_matrix"]
documents = retriever["documents"]


# Retrieval weight configurations
weights = [
    (1.00, 0.00),
    (0.90, 0.10),
    (0.95, 0.05)
]


weight_results = []


for original_weight, intent_weight in weights:

    reciprocal_ranks = []
    hits_at_1 = []
    hits_at_3 = []

    for sample in evaluation_data:

        query = sample["query"]
        intent = sample["intent"]
        expected_id = sample["expected_chunk_id"]

        # Calculate original query similarity
        original_vector = vectorizer.transform([query])

        original_scores = cosine_similarity(
            original_vector,
            document_matrix
        ).flatten()

        # Enrich query using manually assigned intent
        intent_terms = INTENT_SEARCH_TERMS.get(intent, "")

        enriched_query = f"{query} {intent_terms}"

        enriched_vector = vectorizer.transform(
            [enriched_query]
        )

        intent_scores = cosine_similarity(
            enriched_vector,
            document_matrix
        ).flatten()

        # Calculate hybrid retrieval scores
        final_scores = (
            original_weight * original_scores
            + intent_weight * intent_scores
        )

        # Rank documents
        ranked_indices = final_scores.argsort()[::-1][:3]

        ranked_ids = [
            documents[index]["chunk_id"]
            for index in ranked_indices
            if final_scores[index] > 0
        ]

        # Calculate reciprocal rank
        if expected_id in ranked_ids:

            rank = ranked_ids.index(expected_id) + 1

            reciprocal_ranks.append(1 / rank)

        else:
            reciprocal_ranks.append(0)

        # Calculate Precision@1
        hits_at_1.append(
            int(
                bool(ranked_ids)
                and ranked_ids[0] == expected_id
            )
        )

        # Calculate Recall@3
        hits_at_3.append(
            int(expected_id in ranked_ids)
        )

    # Store evaluation metrics
    weight_results.append({
        "original_weight": original_weight,
        "intent_weight": intent_weight,
        "precision_at_1": sum(hits_at_1) / len(evaluation_data),
        "recall_at_3": sum(hits_at_3) / len(evaluation_data),
        "mrr": sum(reciprocal_ranks) / len(evaluation_data)
    })


# Display comparison results
weight_comparison_df = pd.DataFrame(weight_results)

display(
    weight_comparison_df.style.format({
        "precision_at_1": "{:.2%}",
        "recall_at_3": "{:.2%}",
        "mrr": "{:.2%}"
    })
)
````

````python
import pandas as pd
from src.retriever import retrieve_documents

results = []

for item in evaluation_data:
    retrieved = retrieve_documents(
        query=item["query"],
        top_k=3,
        intent=item["intent"]
    )

    retrieved_ids = [
        doc["chunk_id"] for doc in retrieved
    ]

    expected_id = item["expected_chunk_id"]

    correct_rank = (
        retrieved_ids.index(expected_id) + 1
        if expected_id in retrieved_ids
        else None
    )

    results.append({
        "query": item["query"],
        "expected_chunk_id": expected_id,
        "retrieved_chunk_ids": retrieved_ids,
        "correct_rank": correct_rank,
        "precision_at_1": int(
            bool(retrieved_ids)
            and retrieved_ids[0] == expected_id
        ),
        "recall_at_3": int(
            expected_id in retrieved_ids
        ),
        "reciprocal_rank": (
            1 / correct_rank
            if correct_rank is not None
            else 0
        )
    })

results_df = pd.DataFrame(results)

print("RETRIEVAL EVALUATION")
display(results_df)

print("\nSUMMARY METRICS")
print(f"Evaluated queries: {len(results_df)}")
print(f"Precision@1: {results_df['precision_at_1'].mean():.2%}")
print(f"Recall@3: {results_df['recall_at_3'].mean():.2%}")
print(f"MRR: {results_df['reciprocal_rank'].mean():.2%}")

print("\nINCORRECT TOP-1 RESULTS")
display(
    results_df[results_df["precision_at_1"] == 0][
        ["query", "expected_chunk_id", "retrieved_chunk_ids", "correct_rank"]
    ]
)
````

````python
from src.retriever import (
    load_retriever,
    prepare_document_text
)
from sklearn.metrics.pairwise import cosine_similarity

retriever = load_retriever()

vectorizer = retriever["vectorizer"]
documents = retriever["documents"]

query = "Can I cancel my order before it ships?"
query_vector = vectorizer.transform([query])

for chunk_id in ["chunk_003", "chunk_004"]:

    doc = next(
        d for d in documents
        if d["chunk_id"] == chunk_id
    )

    doc_vector = vectorizer.transform([
        prepare_document_text(doc)
    ])

    contributions = (
        query_vector.multiply(doc_vector)
        .toarray()
        .flatten()
    )

    feature_names = vectorizer.get_feature_names_out()

    top_indices = contributions.argsort()[::-1][:10]

    print(f"\n{chunk_id} - {doc['title']}")
    print("Similarity:", cosine_similarity(
        query_vector, doc_vector
    )[0][0])

    print("Top contributing terms:")

    for index in top_indices:
        if contributions[index] > 0:
            print(
                f"{feature_names[index]}: "
                f"{contributions[index]:.4f}"
            )
````

````python
import pandas as pd
import numpy as np

results = []

for query in test_queries:
    message_vector = vectorizer.transform([query])
    scores = classifier.decision_function(message_vector)[0]

    top_indices = np.argsort(scores)[::-1][:3]

    results.append({
        "query": query,
        "top_1_intent": classifier.classes_[top_indices[0]],
        "top_1_score": scores[top_indices[0]],
        "top_2_intent": classifier.classes_[top_indices[1]],
        "top_2_score": scores[top_indices[1]],
        "top_3_intent": classifier.classes_[top_indices[2]],
        "top_3_score": scores[top_indices[2]],
        "margin": scores[top_indices[0]] - scores[top_indices[1]]
    })

confidence_df = pd.DataFrame(results)

display(
    confidence_df.style.format({
        "top_1_score": "{:.4f}",
        "top_2_score": "{:.4f}",
        "top_3_score": "{:.4f}",
        "margin": "{:.4f}"
    })
)
````

````python
import pandas as pd
import joblib
from pathlib import Path
import numpy as np

PROJECT_ROOT = Path.home() / "Desktop" / "Lumora"

EVALUATION_PATH = (
    PROJECT_ROOT
    / "data"
    / "processed"
    / "intent_robustness_evaluation_v1.csv"
)

MODEL_PATH = (
    PROJECT_ROOT
    / "models"
    / "lumora_intent_model_v2.joblib"
)

# Load evaluation data
evaluation_df = pd.read_csv(EVALUATION_PATH)

development_df = evaluation_df[
    evaluation_df["evaluation_partition"] == "development"
].copy()

# Load the saved intent model
intent_artifact = joblib.load(MODEL_PATH)

vectorizer = intent_artifact["vectorizer"]
classifier = intent_artifact["classifier"]

results = []

for _, row in development_df.iterrows():
    query = row["query"]

    query_vector = vectorizer.transform([query])
    scores = classifier.decision_function(query_vector)[0]

    ranked_indices = np.argsort(scores)[::-1]

    predicted_intent = classifier.classes_[ranked_indices[0]]
    second_intent = classifier.classes_[ranked_indices[1]]

    top_score = float(scores[ranked_indices[0]])
    second_score = float(scores[ranked_indices[1]])

    expected_intent = row["expected_intent"]

    if pd.isna(expected_intent):
        intent_correct = None
    else:
        intent_correct = predicted_intent == expected_intent

    results.append({
        "query_id": row["query_id"],
        "query": query,
        "scenario_type": row["scenario_type"],
        "expected_intent": expected_intent,
        "predicted_intent": predicted_intent,
        "second_intent": second_intent,
        "top_score": round(top_score, 4),
        "margin": round(top_score - second_score, 4),
        "intent_correct": intent_correct,
        "expected_action": row["expected_action"]
    })

development_results = pd.DataFrame(results)

display(
    development_results[
        [
            "query_id",
            "query",
            "expected_intent",
            "predicted_intent",
            "second_intent",
            "margin",
            "intent_correct",
            "expected_action"
        ]
    ]
)

print("\nClassification summary:")

labeled_results = development_results.dropna(
    subset=["expected_intent"]
)

print(
    "Labeled queries:",
    len(labeled_results)
)

print(
    "Correct predictions:",
    labeled_results["intent_correct"].sum()
)

print(
    "Accuracy:",
    round(labeled_results["intent_correct"].mean(), 4)
)

print("\nScenario summary:")

display(
    development_results.groupby("scenario_type").agg(
        total_queries=("query_id", "count"),
        labeled_queries=("intent_correct", "count"),
        correct_predictions=("intent_correct", "sum")
    )
)
````
